```python
import numpy as np
import pandas as pd
```

```python
# Set random seed

np.random.seed(101)
```

```python
# Number of employees

n_employees = 600
```

```python
# Generate Employee IDs

emp_id = np.arange(2026,2026 + n_employees)
```

```python
# Departments

departments = np.random.choice(
    ["HR", "Finance", "IT", "Operations", "Sales"],
    size=n_employees
)
```

```python
# Monthly base salary

base_salary = np.random.randint(20000, 90000, size=n_employees)
```

```python
# Bonus amount
bonus = np.random.randint(0, 15000, size=n_employees)
```

```python
# Overtime hours
overtime_hours = np.random.randint(0, 40, size=n_employees)
```

```python
# Leave days taken
leave_days = np.random.randint(0, 10, size=n_employees)
```

```python
# Attendance percentage
attendance_pct = np.random.uniform(60, 100, size=n_employees)
```

```python
# Salary revision percentage
salary_revision_pct = np.random.uniform(-15, 25, size=n_employees)
```

```python
# Payroll risk score logic for employees

risk_score = (
    (salary_revision_pct > 15).astype(int) +
    (attendance_pct < 75).astype(int) +
    (leave_days > 6).astype(int) +
    (overtime_hours > 30).astype(int)
)
```

```python
# Binary payroll risk label for employees

payroll_risk = (risk_score >= 2).astype(int)
```

```python
#create payroll dataframe
employee_payroll_df = pd.DataFrame({
    "Employee_ID": emp_id,
    "Department": departments,
    "Base_Salary": base_salary,
    "Bonus": bonus,
    "Overtime_Hours": overtime_hours,
    "Leave_Days": leave_days,
    "Attendance_Percentage": attendance_pct,
    "Salary_Revision_Percentage": salary_revision_pct,
    "Payroll_Risk": payroll_risk
})

#view dataset
employee_payroll_df.head()
```
